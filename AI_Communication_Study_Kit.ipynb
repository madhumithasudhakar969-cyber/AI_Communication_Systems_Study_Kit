```python
import numpy as np
import matplotlib.pyplot as plt
t=np.linspace(0,1,1000)
fm=5
fc=50
message=np.sin(2*np.pi*fm*t)
carrier=np.sin(2*np.pi*fc*t)
am_signal=(1+message)*carrier
plt.figure(figsize=(10,4))
plt.plot(t,am_signal)
plt.title("Amplitude Modulation(AM)")
plt.xlabel("Time(seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
t=np.linspace(0,1,1000)
fm=5
fc=50
beta=5
message=np.sin(2*np.pi*fm*t)
fm_signal=np.sin(2*np.pi*fc*t+beta*np.sin(2*np.pi*fm*t))
plt.figure(figsize=(10,4))
plt.plot(t,fm_signal)
plt.title("Frequency Modulation(FM)")
plt.xlabel("Time(seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
t=np.linspace(0,1,1000)
fc=10
bits=np.array([1,0,1,1,0])
samples_per_bit=200
digital_bits=np.repeat(bits,samples_per_bit)
time=np.linspace(0,1,len(digital_bits),endpoint=False)
carrier=np.sin(2*np.pi*fc*time)
ask_signal=digital_bits*carrier
plt.figure(figsize=(10,4))
plt.plot(time,ask_signal)
plt.title("Amplitude Shift Keying(ASK)")
plt.xlabel("Time(seconds)")
plt.ylabel("")

```
